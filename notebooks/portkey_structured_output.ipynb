```python
import os
from pprint import pprint
from datetime import date
from portkey_ai import Portkey
from typing import Optional
from pydantic import BaseModel, Field

portkey = Portkey(
    base_url="https://ai-gateway.apps.cloud.rt.nyu.edu/v1/",
    api_key=os.getenv("PORTKEY_API_KEY"),
)
```

```python
class ArticleSpec(BaseModel):
    """Information about an article written in markdown"""

    title: str = Field(description="Article title")
    author: str = Field(description="Name of the author")
    publication_date: Optional[date] = Field(description="Date written in YYYY-MM-DD format.")

prompt = prompt = """
Extract data from the following text:

<text>
# Structured Data
By Carson Sievert
</text>
"""
```

### Enforcing response format

```python
response = portkey.responses.parse(
    model="@vertexai/gemini-3-flash-preview",
    input=prompt,
    text_format=ArticleSpec
)
pprint(response.output_parsed)
```
