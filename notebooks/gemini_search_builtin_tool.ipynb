```python
import os
from rich.console import Console
from rich.markdown import Markdown
from portkey_ai import Portkey

portkey = Portkey(
    base_url="https://ai-gateway.apps.cloud.rt.nyu.edu/v1/",
    api_key=os.getenv("PORTKEY_API_KEY"),
)
```

### Run the prompt without the tool

```python
response = portkey.responses.create(
    model="@vertexai/gemini-2.5-flash",
    input="Tell me about the latest HPC cluster at NYU",
    max_output_tokens=8196,
)

console = Console()
console.print(Markdown(response.output_text))
```

### Run the prompt with the tool

```python
response = portkey.responses.create(
    model="@vertexai/gemini-2.5-flash",
    tools=[{"type": "function", "name": "google_search"}],
    input="Tell me about the latest HPC cluster at NYU",
    max_output_tokens=8196,
)
console = Console()
console.print(Markdown(response.output_text))
```
