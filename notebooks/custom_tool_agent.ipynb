```python
import os
import json
import random

from IPython.display import display, Markdown

from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIResponsesModel
from pydantic_ai.providers.openai import OpenAIProvider
```

#### Define a python function and specification for passing it to the Agent

```python
agent = Agent(
    name="agent-with-custom-tool",
    model=OpenAIResponsesModel(
        model_name="@vertexai/gemini-3.7-flash",
        provider=OpenAIProvider(
            base_url="https://ai-gateway.apps.cloud.rt.nyu.edu/v1/",
            api_key=os.getenv("PORTKEY_API_KEY"),
        ),
    ),
)

@agent.tool_plain  
def roll_dice(N: int) -> int:
  """Roll a N-sided die and return the result."""
  return random.randint(1, N)
```

```python
result = await agent.run('Roll a 10 sided die and check if the dice roll was valid. Explain your reasoning.') # await is added because the agent is run asynchronously
```

```python
display(Markdown(result.output))
```
